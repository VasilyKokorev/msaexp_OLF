```python
import yaml
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Read the file in

file_name = '/Users/vasily/Documents/projects/msa/valentino/overdensity/spectra/corr/output/12133119_prism.yaml'

with open(file_name, 'r') as file:
        spec = yaml.safe_load(file)
```

```python
# Everything the file contains
spec.keys()
```

```python
# Plot the best fit spectrum
# Units are flambda, erg/s/cm-2/A x 1e-20

fig = plt.figure(figsize=(8,5))
ax = fig.add_subplot(111)

ax.step(spec['wave'],spec['flam'],where='mid',color='k',label='spec')
ax.step(spec['wave'],spec['model'],where='mid',color='r',label='model')

# If you just want to separate the fit to lines alone
ax.step(spec['wave'],spec['mline'],where='mid',color='royalblue',label='lines')

ax.legend()
ax.grid(alpha=.3)
```

```python
# Some params
# FWHM of lines, in km/s, rest frame
print('Narrow width',spec['fwhm_narrow'],'km/s')
print('Broad width',spec['fwhm_broad'],'km/s')
```
